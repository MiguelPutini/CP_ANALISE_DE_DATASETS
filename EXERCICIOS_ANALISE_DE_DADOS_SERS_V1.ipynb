### Dataset 3 — Power Consumption of Tetouan City (UCI)

Fonte: https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city

Situação

Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.

```python
import pandas as pd

# 1. Carregue o dataset (usando o arquivo arrumado)
df = pd.read_csv('Datset3_2.csv')

# Em vez de usar o rename (que falha se tiver um espaço a mais),
# vamos forçar o nome das 6 colunas na ordem exata da sua imagem:
df.columns = ['Temperature', 'Humidity', 'Wind Speed', 'Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']

# 2. Determine o consumo máximo registrado em cada uma das três zonas
max_z1 = df['Consumo_Zona_1'].max()
max_z2 = df['Consumo_Zona_2'].max()
max_z3 = df['Consumo_Zona_3'].max()

print("--- Passo 2: Consumo Máximo por Zona ---")
print(f"Zona 1: {max_z1:.2f}")
print(f"Zona 2: {max_z2:.2f}")
print(f"Zona 3: {max_z3:.2f}\n")

# 3. Identifique qual zona apresenta o maior pico de consumo na amostra
zonas_maximos = {
    'Consumo_Zona_1': max_z1,
    'Consumo_Zona_2': max_z2,
    'Consumo_Zona_3': max_z3
}

zona_pico = max(zonas_maximos, key=zonas_maximos.get)
valor_pico = zonas_maximos[zona_pico]

print("--- Passo 3: Zona de Maior Pico ---")
print(f"A zona com o maior pico de consumo na amostra é a {zona_pico}, atingindo {valor_pico:.2f}\n")

# 4. Para a zona identificada, calcule 70% do valor máximo e crie um DataFrame com os registros acima desse limiar
limiar_70 = valor_pico * 0.70
df_acima_limiar = df[df[zona_pico] > limiar_70]

print("--- Passo 4 & 5: Filtro de 70% do Máximo ---")
print(f"O limiar de 70% para a {zona_pico} é de {limiar_70:.2f}")

# 5. Conte os registros e calcule o percentual que eles representam
total_registros = len(df)
qtd_acima_limiar = len(df_acima_limiar)
percentual_acima = (qtd_acima_limiar / total_registros) * 100

print(f"Quantidade de registros acima desse limiar: {qtd_acima_limiar}")
print(f"Isso representa {percentual_acima:.2f}% do total de registros da amostra.\n")

# 6. Calcule a temperatura média e crie um segundo DataFrame (consumo acima do limiar e temperatura acima da média)
temp_media = df['Temperature'].mean()
df_temp_alta = df_acima_limiar[df_acima_limiar['Temperature'] > temp_media]

qtd_final = len(df_temp_alta)
print("--- Passo 6: Inclusão da Variável Ambiental ---")
print(f"A temperatura média da amostra é {temp_media:.2f}°C")
print(f"Registros simultâneos (Consumo > 70% DO PICO e Temperatura > MÉDIA): {qtd_final}\n")

# 7. Compare os dois conjuntos e explique o que ocorreu com a quantidade de registros
print("--- Passo 7: Análise Comparativa ---")
print(f"Ao adicionar a condição ambiental, a quantidade de registros caiu de {qtd_acima_limiar} para {qtd_final}.")

if qtd_final > (qtd_acima_limiar / 2):
    print("Explicação: A redução não foi tão drástica, indicando que a grande maioria dos picos de "
          "consumo ocorre justamente em momentos de temperatura acima da média. Isso sugere uma forte "
          "relação entre o aumento do calor e o pico de gasto energético (ex: uso de refrigeração).")
else:
    print("Explicação: Houve uma redução significativa nos registros. Isso indica que muitos dos momentos "
          "de altíssimo consumo de energia não dependem de temperaturas elevadas, podendo estar associados a "
          "horários de pico residencial/industrial independentes do clima.")
```

```python

df_solar = df5[df5['Source'] == 'Solar'].copy().rename(columns={'Production': 'Geracao_Solar'})
df_eolica = df5[df5['Source'] == 'Wind'].copy().rename(columns={'Production': 'Geracao_Eolica'})
```

```python

print("\n--- Comparação de Frequência ---")
if perc_solar > perc_eolica:
    print("A fonte SOLAR aparece com maior frequência proporcional operando acima de 70% do seu máximo.")
elif perc_eolica > perc_solar:
    print("A fonte EÓLICA aparece com maior frequência proporcional operando acima de 70% do seu máximo.")
else:
    print("Ambas as fontes possuem a mesma frequência proporcional de registros acima de 70% dos máximos.")
```

### 7. Por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes?

Fontes renováveis diferentes operam em escalas estruturais e capacidades nominais totalmente distintas. Se você aplicasse um valor absoluto fixo (por exemplo, exigir um corte numérico idêntico para ambas), criaria uma distorção grave na análise: para a usina menor, o número poderia representar um pico inalcançável de 90% da sua capacidade máxima, enquanto para a outra representaria um patamar comum de operação. Utilizar 70% do máximo individual normaliza essas diferenças, permitindo avaliar justamente o desempenho relativo de cada tecnologia frente ao seu próprio teto de produção.

### Dataset 6 — Individual Household Electric Power Consumption (UCI)

Fonte: https://archive.ics.uci.edu/dataset/235/individual+household+electric+power+consumption

Situação

Uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio. O dataset é o mesmo utilizado como referência em aula, mas o critério de análise será diferente.

```python
import pandas as pd
from IPython.display import display

# ==============================================================================
# 1. Carregue a nova amostra e organize os atributos utilizando nomes mais simples
# ==============================================================================
dados = pd.read_csv("SAMPLE_ENERGY_DATA.csv")

dados = dados.rename(columns={
    "Global_active_power": "Pot_Ativa",
    "Global_reactive_power": "Pot_Reativa",
    "Voltage": "Tensao",
    "Global_intensity": "Corrente",
    "Sub_metering_1": "Sub_1",
    "Sub_metering_2": "Sub_2",
    "Sub_metering_3": "Sub_3"
})

print("--- 1. ESTRUTURA E DADOS DA AMOSTRA ---")
dados.info()
print(f"\nDimensões do dataset original: {dados.shape}")
display(dados.head())

# ==============================================================================
# 2. Determine o valor máximo de potência ativa
# ==============================================================================
max_potencia = dados['Pot_Ativa'].max()

print("\n--- 2. POTÊNCIA ATIVA MÁXIMA ---")
print(f"O valor máximo de potência ativa é: {max_potencia}")

# ==============================================================================
# 3. Calcule 75% do valor máximo e crie um DataFrame acima desse limite
# ==============================================================================
limiar_75 = 0.75 * max_potencia
df_alto_consumo = dados[dados['Pot_Ativa'] > limiar_75]

# ==============================================================================
# 4. Calcule a quantidade e o percentual de registros selecionados
# ==============================================================================
qtd_alto = len(df_alto_consumo)
percentual = (qtd_alto / len(dados)) * 100

print("\n--- 3 & 4. REGISTROS ACIMA DE 75% DA POTÊNCIA ---")
print(f"Limiar de corte (75%): {limiar_75}")
print(f"Quantidade de registros: {qtd_alto}")
print(f"Percentual da amostra: {percentual:.3f}%")

# ==============================================================================
# 5. Calcule a corrente média da amostra
# ==============================================================================
corrente_media = dados['Corrente'].mean()

print("\n--- 5. CORRENTE MÉDIA ---")
print(f"Corrente média da amostra: {corrente_media:.3f} amperes")

# ==============================================================================
# 6. Crie um 2º DataFrame contendo potência > 75% E corrente > média
# ==============================================================================
df_duplo_filtro = dados[(dados['Pot_Ativa'] > limiar_75) & (dados['Corrente'] > corrente_media)]
qtd_final = len(df_duplo_filtro)

print("\n--- 6. FILTRO SIMULTÂNEO (POTÊNCIA E CORRENTE) ---")
print(f"Registros encontrados com as duas condições: {qtd_final}")
print(f"Dimensões do novo DataFrame: {df_duplo_filtro.shape}")
display(df_duplo_filtro.head())

# ==============================================================================
# 7. Compare os dois conjuntos e explique o efeito da inclusão da corrente
# ==============================================================================
print("\n--- 7. COMPARAÇÃO E EXPLICAÇÃO ---")
print(f"Registros do 1º filtro (Só Potência): {qtd_alto}")
print(f"Registros do 2º filtro (Potência + Corrente): {qtd_final}")

if qtd_alto == qtd_final:
    print("\nExplicação: A quantidade de registros não se alterou. O efeito de incluir "
          "a corrente como segunda condição foi nulo porque, fisicamente, potência e corrente "
          "são grandezas diretamente proporcionais. Todos os momentos de altíssimo pico de energia "
          "(potência ativa) exigiram, naturalmente, uma corrente muito acima da média.")
else:
    print("\nExplicação: Houve alteração no número de registros. Apenas uma parte dos "
          "picos de potência apresentou corrente acima da média.")
```
